```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("Add 5:", arr + 5)
print("Multiply by 2:", arr * 2)
print("Squared:", arr ** 2)
```

```python
# Loop method
loop_result = []

for x in arr:
    loop_result.append(int(x * 2))

print("Loop result:", loop_result)

# Vectorized method
vectorized_result = arr * 2

print("Vectorized result:", vectorized_result)
```

**Comparison:** Both methods produce the same result, but the vectorized NumPy approach is more concise and is generally more efficient for large arrays.
