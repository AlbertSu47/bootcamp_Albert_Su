# Homework Starter — Stage 6: Data Preprocessing
Use this notebook to apply your cleaning functions and save processed data.

```python
import pandas as pd

from src import cleaning
```

## Load Raw Dataset

```python
df = pd.read_csv('data/raw/sample_data.csv')
df.head()
```

## Apply Cleaning Functions

```python
# Keep the original data for comparison
df_original = df.copy()

# 1. Fill missing numeric values with the median
df_clean = cleaning.fill_missing_median(
    df,
    ["age", "income", "score"]
)

# 2. Drop columns with more than 50% missing values
df_clean = cleaning.drop_missing(
    df_clean,
    threshold=0.5
)

# 3. Normalize selected numeric columns to [0, 1]
df_clean = cleaning.normalize_data(
    df_clean,
    ["age", "income", "score"]
)

df_clean
```

## Save Cleaned Dataset

```python
output_path = "data/processed/sample_data_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Saved cleaned dataset to:", output_path)
```

## Original vs Cleaned Data

Compare the original dataset with the cleaned dataset to verify the preprocessing steps.

```python
print("Original shape:", df_original.shape)
print("Cleaned shape:", df_clean.shape)

print("\nOriginal missing values:")
print(df_original.isna().sum())

print("\nCleaned missing values:")
print(df_clean.isna().sum())

print("\nOriginal numeric summary:")
display(df_original[["age", "income", "score"]].describe())

print("\nCleaned numeric summary:")
display(df_clean[["age", "income", "score"]].describe())
```

## Cleaning Strategy and Assumptions

- Missing values in `age`, `income`, and `score` are filled with the median because these are numeric variables and the median is less sensitive to extreme values.
- Columns with more than 50% missing values are dropped because they contain too little information to be reliably used.
- `age`, `income`, and `score` are normalized to the range [0, 1] so that variables measured on different scales can be compared more consistently.
- `zipcode` and `city` are left unchanged because they are categorical or identifier-like variables rather than continuous numeric features.
- The original dataset is preserved separately as `df_original` so that preprocessing changes can be compared and reviewed.
