The goal of this project is to evaluate whether a simple ETF trading strategy is likely to remain effective in the future. Historical market data will be used to analyze the strategy's performance, risk, and consistency over time. The evaluation will consider expected performance and potential risks rather than relying only on historical returns, including measures such as return, volatility, drawdown, and risk-adjusted performance.

The primary stakeholder is a portfolio manager or investor who must decide whether to continue using the ETF strategy, while a quantitative analyst or investment analyst will run the analysis and report the results. ## Stakeholder Artifact

### Stakeholder Memo — ETF Strategy Evaluation

**To:** Portfolio Manager / Investor  
**From:** Quantitative Analyst  
**Subject:** Evaluation of ETF Strategy Sustainability  
**Stage:** Problem Framing & Scoping (Stage 01)

### Decision Context

The key decision is whether the current ETF trading strategy should continue to be used in the future. Historical performance alone is not sufficient for this decision because market conditions, transaction costs, and risk characteristics may change over time.

The analysis will therefore evaluate both the potential return and the associated risk of the strategy.

### Stakeholder Need

The portfolio manager needs a clear and practical assessment of whether the strategy is likely to remain useful under realistic market conditions.

The analysis should help answer:

- Is the strategy expected to remain profitable?
- How much risk is associated with the strategy?
- Is performance consistent enough to justify continued use?
- Should the strategy be continued, modified, or discontinued?

### Proposed Analysis

Historical ETF market data will be analyzed using performance and risk measures including:

- Expected return
- Volatility
- Drawdown
- Risk-adjusted performance

The analysis will also consider transaction costs, liquidity, and changes in market conditions.

### Key Risks and Limitations

Past performance may not accurately represent future results. The strategy may also be affected by structural market changes, overfitting, transaction costs, slippage, and the choice of benchmark or evaluation period.

### Intended Deliverable

The final deliverable will be a concise performance and risk assessment with a recommendation to:

- Continue the strategy,
- Modify the strategy, or
- Discontinue the strategy.
